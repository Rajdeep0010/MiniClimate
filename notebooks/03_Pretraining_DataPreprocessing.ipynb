```python
# CHANGE THIS PATH (IMPORTANT)
root_path = "/content/drive/MyDrive/Rajdeep_Document/data/Climate_change_dataset-20260422T121056Z-3-001/Climate_change_dataset"

print("Path exists:", os.path.exists(root_path))

all_rows = []

# Known useful column names (edit if needed)
TEXT_COLUMNS_HINT = ["text", "post", "content", "caption", "message"]

def extract_text_rows(df, file_name, folder_name):
    rows = []

    for col in df.columns:
        try:
            col_data = df[col].dropna().astype(str)

            # Skip numeric-like columns
            if col_data.str.match(r'^\d+$').all():
                continue

            # Skip very short avg text (IDs, codes)
            if col_data.str.len().mean() < 5:
                continue

            for val in col_data:
                rows.append({
                    "text": val,
                    "source_file": file_name,
                    "folder": folder_name,
                    "column": col
                })

        except Exception as e:
            continue

    return rows


# DEBUG: Check files being found
print("\nScanning files...\n")

file_counter = 0

for root, dirs, files in os.walk(root_path):
    print(f"📁 {root} -> {len(files)} files")

    folder_name = os.path.basename(root)

    for file in tqdm(files):
        file_path = os.path.join(root, file)
        file_counter += 1

        try:
            if file.endswith(".csv"):
                try:
                    df = pd.read_csv(file_path, encoding='utf-8', on_bad_lines='skip')
                except:
                    df = pd.read_csv(file_path, encoding='latin1', on_bad_lines='skip')

            elif file.endswith(".xlsx"):
                df = pd.read_excel(file_path)

            else:
                continue

            print(f"\n Processing: {file} | Rows: {len(df)}")
            print("Columns:", list(df.columns))

            extracted = extract_text_rows(df, file, folder_name)

            print(f"Extracted rows: {len(extracted)}")

            all_rows.extend(extracted)

        except Exception as e:
            print(f"Skipping {file}: {e}")


print("\nTotal files processed:", file_counter)
print("Total extracted rows:", len(all_rows))


# STOP if nothing extracted
if len(all_rows) == 0:
    raise ValueError("No data extracted. Check path or column logic.")


# Create DataFrame
df_all = pd.DataFrame(all_rows)

print("\nDataFrame created")
print(df_all.head())
print(df_all.info())
```

```python
import os
import re
import json
import hashlib
import unicodedata

import numpy as np
import pandas as pd
import ftfy
from tqdm import tqdm
from langdetect import detect, LangDetectException

# ── Config ───────────────────────────────────────────────────
INPUT_FILENAME   = "/content/drive/MyDrive/Rajdeep_Document/data/climate_dataset_full.csv"
OUTPUT_DIR       = "/content/drive/MyDrive/Rajdeep_Document/data"
os.makedirs(OUTPUT_DIR, exist_ok=True)

MIN_WORDS         = 5
MAX_WORDS         = 512
MIN_CLIMATE_SCORE = 1     # Keep: any text scoring >= 1
TRAIN_RATIO       = 0.90
VAL_RATIO         = 0.05
```

```python
# ============================================================
# STAGE 1 — LOAD & SNAPSHOT
# ============================================================
print("\n" + "="*60)
print(" STAGE 1 — Loading Raw Dataset")
print("="*60)

df_raw = pd.read_csv(INPUT_FILENAME)
df     = df_raw.copy()

before = {
    "total_rows"            : len(df),
    "columns"               : list(df.columns),
    "null_counts"           : df.isnull().sum().to_dict(),
    "duplicate_texts"       : int(df["text"].duplicated().sum()),
    "avg_text_len_chars"    : round(df["text"].astype(str).str.len().mean(), 1),
    "source_files"          : int(df["source_file"].nunique()),
    "folders"               : df["folder"].value_counts().to_dict(),
}

print(f"  Rows loaded      : {before['total_rows']:,}")
print(f"  Columns          : {before['columns']}")
print(f"  Duplicate texts  : {before['duplicate_texts']:,}")
print(f"  Avg text length  : {before['avg_text_len_chars']} chars")
print(f"  Folder breakdown : {before['folders']}")
```

```python
# ============================================================
# STAGE 2 — SCHEMA NORMALIZATION
# ============================================================
print("\n" + "="*60)
print(" STAGE 2 — Schema Normalization")
print("="*60)

if "column" in df.columns:
    df.drop(columns=["column"], inplace=True)  # always-zero, no value

df = df[["text", "source_file", "folder"]].copy()
df["text"]        = df["text"].astype(str).str.strip()
df["source_file"] = df["source_file"].astype(str)
df["folder"]      = df["folder"].astype(str)

before_null = len(df)
df = df[df["text"].str.len() > 0].reset_index(drop=True)
print(f"  Columns kept      : {list(df.columns)}")
print(f"  Empty rows dropped: {before_null - len(df):,}")
```

```python
# ============================================================
# STAGE 2 — SCHEMA NORMALIZATION
# ============================================================
print("\n" + "="*60)
print(" STAGE 2 — Schema Normalization")
print("="*60)

# Drop the always-zero 'column' field if present
if "column" in df.columns:
    df.drop(columns=["column"], inplace=True)

df = df[["text", "source_file", "folder"]].copy()
df["text"]        = df["text"].astype(str).str.strip()
df["source_file"] = df["source_file"].astype(str)
df["folder"]      = df["folder"].astype(str)

# Drop rows where text is empty after stripping
before_null = len(df)
df = df[df["text"].str.len() > 0].reset_index(drop=True)
print(f"  Schema reduced to : {list(df.columns)}")
print(f"  Empty rows dropped: {before_null - len(df):,}")
```

```python
# ============================================================
# STAGE 3 — UNICODE REPAIR
# ============================================================
print("\n" + "="*60)
print(" STAGE 3 — Unicode Repair (ftfy)")
print("="*60)

df["text"] = df["text"].apply(lambda x: ftfy.fix_text(str(x)))
print(f"  ftfy encoding fix applied to all {len(df):,} rows.")
```

```python
# ============================================================
# STAGE 4 — DEDUPLICATION
# ============================================================
print("\n" + "="*60)
print(" STAGE 4 — Deduplication")
print("="*60)

before_dedup = len(df)

# 4a. Exact
df = df.drop_duplicates(subset="text", keep="first").reset_index(drop=True)
exact_removed = before_dedup - len(df)
print(f"  Exact duplicates removed : {exact_removed:,}")

# 4b. Near-duplicates via normalised fingerprint
def fingerprint(text: str) -> str:
    t = re.sub(r'\s+', ' ', text.lower()).strip()
    t = re.sub(r'[^\w\s]', '', t)
    return hashlib.md5(t.encode('utf-8')).hexdigest()

df["_fp"]    = df["text"].apply(fingerprint)
before_near  = len(df)
df           = df.drop_duplicates(subset="_fp", keep="first").reset_index(drop=True)
near_removed = before_near - len(df)
df.drop(columns=["_fp"], inplace=True)

print(f"  Near-duplicates removed  : {near_removed:,}")
print(f"  Rows after dedup         : {len(df):,}")
```

```python
# ============================================================
# STAGE 5 — FAN / SPAM COMMENT REMOVAL
# ============================================================
print("\n" + "="*60)
print(" STAGE 5 — Fan/Spam Comment Removal")
print("="*60)

FAN_PATTERNS = re.compile(
    r'^(nice|good|great|best|superb|amazing|excellent|wonderful|'
    r'first\s*comment|love\s*you|love\s*it|very\s*nice|'
    r'well\s*done|thank\s*you\s*(so\s*much)?|'
    r'bhai\s*(great|nice|good|aap)|'
    r'sir\s*(ji)?\s*(nice|great|good|aap|please)|'
    r'subscribe|like\s*karo|share\s*karo|'
    r'[❤👍🙏]{3,}|[!]{4,}|[.]{4,})\b',
    re.IGNORECASE | re.UNICODE
)

def is_fan_spam(text: str) -> bool:
    stripped = text.strip()
    # Pure emoji/punctuation rows
    if not re.search(r'[a-zA-Z\u0900-\u097F\u0600-\u06FF]', stripped):
        return True
    # Matches fan patterns and is short (< 10 words)
    if FAN_PATTERNS.search(stripped) and len(stripped.split()) < 10:
        return True
    return False

before_fan = len(df)
df = df[~df["text"].apply(is_fan_spam)].reset_index(drop=True)
fan_removed = before_fan - len(df)
print(f"  Fan/spam rows removed : {fan_removed:,}")
print(f"  Rows remaining        : {len(df):,}")
```

```python
# ============================================================
# STAGE 6 — NOISE REMOVAL
# ============================================================
print("\n" + "="*60)
print(" STAGE 6 — Noise Removal")
print("="*60)

def is_noisy(text: str) -> bool:
    # Must contain at least one letter from any major script
    has_letter = bool(re.search(
        r'[a-zA-Z'
        r'\u0900-\u097F'    # Devanagari (Hindi, Marathi, Sanskrit)
        r'\u0980-\u09FF'    # Bengali
        r'\u0A80-\u0AFF'    # Gujarati
        r'\u0B00-\u0B7F'    # Odia
        r'\u0C00-\u0C7F'    # Telugu
        r'\u0C80-\u0CFF'    # Kannada
        r'\u0D00-\u0D7F'    # Malayalam
        r'\u0600-\u06FF'    # Arabic / Urdu
        r'\u0400-\u04FF'    # Cyrillic
        r'\u4E00-\u9FFF'    # CJK
        r'\uAC00-\uD7A3'    # Korean
        r']', text
    ))
    if not has_letter:
        return True
    # URL-only content
    url_ratio = len(re.findall(r'https?://\S+|www\.\S+', text)) * 20 / max(len(text), 1)
    if url_ratio > 0.5:
        return True
    # Single repeated character
    if re.fullmatch(r'(.)\1{9,}', text.replace(' ', '')):
        return True
    return False

before_noise = len(df)
df = df[~df["text"].apply(is_noisy)].reset_index(drop=True)
noise_removed = before_noise - len(df)
print(f"  Noise rows removed : {noise_removed:,}")
print(f"  Rows remaining     : {len(df):,}")
```

```python
# ============================================================
# STAGE 7 — TEXT CLEANING & NORMALIZATION
# ============================================================
print("\n" + "="*60)
print(" STAGE 7 — Text Cleaning & Normalization")
print("="*60)

def clean_text(text: str) -> str:
    text = (text.replace("&amp;", "&").replace("&lt;", "<")
                .replace("&gt;", ">").replace("&quot;", '"')
                .replace("&#39;", "'"))
    text = re.sub(r'<[^>]+>', ' ', text)
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    text = re.sub(r'@\w+', '', text)
    text = re.sub(r'#(\w+)', r'\1', text)
    # Remove emojis
    emoji_re = re.compile(
        "[\U0001F600-\U0001F64F\U0001F300-\U0001F5FF"
        "\U0001F680-\U0001F6FF\U0001F1E0-\U0001F1FF"
        "\U00002702-\U000027B0\U000024C2-\U0001F251]+",
        flags=re.UNICODE
    )
    text = emoji_re.sub('', text)
    text = ''.join(
        c for c in text
        if unicodedata.category(c)[0] != 'C' or c in '\n\t '
    )
    text = re.sub(r'([!?.]){2,}', r'\1', text)
    text = re.sub(r'\r\n|\r', '\n', text)
    text = re.sub(r'\n{3,}', '\n\n', text)
    text = re.sub(r'[ \t]{2,}', ' ', text)
    return text.strip()

def normalize_text(text: str) -> str:
    text = unicodedata.normalize("NFC", text)  # safe for ALL scripts
    text = (text.replace('\u2019', "'").replace('\u2018', "'")
                .replace('\u201c', '"').replace('\u201d', '"')
                .replace('\u2013', '-').replace('\u2014', ' - ')
                .replace('\u00a0', ' '))
    # Scientific notation normalization
    text = re.sub(r'(\d+)\s*°\s*[Cc]', r'\1°C', text)
    text = re.sub(r'(\d+)\s*°\s*[Ff]', r'\1°F', text)
    text = re.sub(r'CO\s*2\b',    'CO2', text, flags=re.IGNORECASE)
    text = re.sub(r'CH\s*4\b',    'CH4', text, flags=re.IGNORECASE)
    text = re.sub(r'N\s*2\s*O\b', 'N2O', text, flags=re.IGNORECASE)
    text = re.sub(r'\s+([.,;:!?])', r'\1', text)
    return text.strip()

df["text_clean"] = df["text"].apply(clean_text).apply(normalize_text)

before_empty = len(df)
df = df[df["text_clean"].str.strip().str.len() > 0].reset_index(drop=True)
print(f"  Rows emptied by cleaning : {before_empty - len(df):,}")
print(f"  Rows after cleaning      : {len(df):,}")
```

```python
# ============================================================
# STAGE 8 — LENGTH & QUALITY FILTERING
# ============================================================
print("\n" + "="*60)
print(" STAGE 8 — Length & Quality Filtering")
print("="*60)

df["word_count"] = df["text_clean"].str.split().str.len()
before_len       = len(df)
df               = df[df["word_count"] >= MIN_WORDS].reset_index(drop=True)
length_removed   = before_len - len(df)

def truncate_words(text: str, max_w: int = MAX_WORDS) -> str:
    words = text.split()
    return ' '.join(words[:max_w]) if len(words) > max_w else text

df["text_clean"] = df["text_clean"].apply(truncate_words)
df["word_count"] = df["text_clean"].str.split().str.len()

print(f"  Too-short removed (< {MIN_WORDS} words) : {length_removed:,}")
print(f"  Rows after length filter        : {len(df):,}")

# Quality: >= 30% word-tokens (lowered from 50% to support non-Latin scripts)
def quality_score(text: str) -> float:
    tokens = text.split()
    if not tokens:
        return 0.0
    alpha = sum(1 for t in tokens if re.search(r'\w', t) and not t.isdigit())
    return alpha / len(tokens)

df["quality"]   = df["text_clean"].apply(quality_score)
before_qual     = len(df)
df              = df[df["quality"] >= 0.30].reset_index(drop=True)
qual_removed    = before_qual - len(df)
print(f"  Low-quality removed (< 30%)     : {qual_removed:,}")
print(f"  Rows after quality filter        : {len(df):,}")
```

```python
# STAGE 9 — LANGUAGE DETECTION
print("\n" + "="*60)
print(" STAGE 9 — Language Detection (tag all, keep all)")
print("="*60)

def safe_detect(text: str) -> str:
    try:
        return detect(str(text))
    except LangDetectException:
        return "unknown"

tqdm.pandas(desc="  Detecting languages")
df["lang"] = df["text"].progress_apply(safe_detect)

print(f"\n  Top languages:")
for lang, cnt in df["lang"].value_counts().head(12).items():
    print(f"    {lang:>10} : {cnt:>6,}  ({cnt/len(df)*100:.1f}%)")
print(f"  Total unique languages : {df['lang'].nunique()}")
```

```python
# ============================================================
# STAGE 10 — CLIMATE RELEVANCE SCORING
# ============================================================
# KEY FIX: Comprehensive multilingual keyword list.
# Previous list missed ~4,700 Hindi tree/nature texts and
# most Hinglish climate content.
print("\n" + "="*60)
print(" STAGE 10 — Climate Relevance Scoring")
print("="*60)

CLIMATE_KEYWORDS = {

    # Score +3 — Core climate science & policy terms
    "high": [
        # English
        "climate change", "global warming", "greenhouse gas", "greenhouse effect",
        "carbon dioxide", "co2 emissions", "carbon footprint", "carbon neutral",
        "net zero", "ipcc", "paris agreement", "kyoto protocol", "decarbonization",
        "fossil fuels", "fossil fuel", "sea level rise", "ice cap melting",
        "glacier melt", "glaciers melting", "extreme weather", "heat wave",
        "ocean acidification", "methane emissions", "nitrous oxide", "ozone layer",
        "carbon sequestration", "carbon tax", "emission trading", "tipping point",
        "feedback loop", "albedo effect", "permafrost thaw", "arctic melt",
        "biodiversity loss", "sixth extinction", "mass extinction",
        # Hindi — Devanagari
        "जलवायु परिवर्तन", "ग्लोबल वार्मिंग", "ग्रीनहाउस गैस",
        "कार्बन उत्सर्जन", "जीवाश्म ईंधन", "ग्लेशियर पिघल",
        "समुद्र स्तर", "ओजोन परत", "कार्बन फुटप्रिंट",
        "नवीकरणीय ऊर्जा", "पेरिस समझौता", "जैव विविधता",
        # Hinglish / Romanized Hindi
        "jalvayu parivartan", "global warming ki", "greenhouse gas",
        "carbon footprint", "fossil fuel", "nvikaraniya urja",
        "paris samjhauta", "jalvayu sankat",
    ],

    # Score +2 — Strong climate/environment indicators
    "medium": [
        # English
        "temperature rise", "rising temperature", "climate crisis",
        "climate emergency", "climate action", "climate policy",
        "renewable energy", "clean energy", "solar power", "wind power",
        "wind energy", "electric vehicle", "deforestation", "afforestation",
        "reforestation", "flood", "drought", "wildfire", "cyclone",
        "hurricane", "typhoon", "storm surge", "heatwave", "pollution",
        "air pollution", "water pollution", "plastic pollution",
        "carbon emissions", "emissions", "sustainability", "sustainable",
        "ecology", "ecosystem", "habitat destruction", "coral reef",
        "ice sheet", "polar ice", "arctic", "antarctic", "amazon",
        "rainforest", "forest fire", "forest cover", "tree cover",
        "tree plantation", "plant trees", "save trees", "cut trees",
        "cutting trees", "ban deforestation", "recycle", "zero waste",
        "clean water", "water scarcity", "water crisis", "food security",
        "natural disaster", "disaster relief",
        # Hindi — Devanagari
        "प्रदूषण", "पर्यावरण", "वनों की कटाई", "पेड़ लगाओ",
        "पेड़ बचाओ", "वृक्षारोपण", "जल संकट", "वायु प्रदूषण",
        "जल प्रदूषण", "बाढ़", "सूखा", "तूफान", "गर्मी",
        "तापमान", "ग्लेशियर", "हिमनद", "समुद्र", "महासागर",
        "जंगल", "जंगल की आग", "जंगल कटाई", "प्राकृतिक आपदा",
        "जल संरक्षण", "सौर ऊर्जा", "पवन ऊर्जा", "हरित ऊर्जा",
        "धरती बचाओ", "पृथ्वी बचाओ", "पर्यावरण बचाओ",
        "जैव विविधता", "वन्यजीव", "प्रकृति",
        # Hinglish / Romanized
        "pradushan", "paryavaran", "ped lagao", "ped bachao",
        "jungle katai", "jal sankat", "baadh", "sukha", "toofan",
        "tapman", "glacier", "samudra", "jungle", "jungle ki aag",
        "dharti bachao", "prithvi bachao", "prakriti bachao",
        "harit urja", "soura urja", "pawan urja", "swachh urja",
        "jal pradushan", "vayu pradushan", "mitti pradushan",
    ],

    # Score +1 — Contextual / broad environment terms
    "low": [
        # English
        "temperature", "weather", "climate", "environment", "nature",
        "green", "earth", "planet", "trees", "tree", "forest", "jungle",
        "ocean", "river", "water", "air", "soil", "energy", "electricity",
        "coal", "oil", "gas", "nuclear", "industry", "factory",
        "agriculture", "farming", "crops", "food", "species", "animal",
        "wildlife", "habitat", "conservation", "protect", "save",
        "future generation", "our planet", "our earth", "save earth",
        "save planet", "go green", "zero emission", "low carbon",
        "smog", "haze", "dust", "toxic", "chemical", "waste",
        "landfill", "composting", "organic", "biodegradable",
        "protest", "activist", "movement", "greta", "scientist",
        "research", "study", "data", "report", "model", "prediction",
        # Hindi — Devanagari
        "पेड़", "वृक्ष", "नदी", "पानी", "वायु", "मिट्टी",
        "ऊर्जा", "बिजली", "कोयला", "तेल", "गैस", "उद्योग",
        "खेती", "फसल", "खाना", "पशु", "वन्यजीव", "संरक्षण",
        "भविष्य", "आने वाली पीढ़ी", "हमारी धरती", "हमारी पृथ्वी",
        "बचाना", "रक्षा", "हरित", "स्वच्छ", "शुद्ध",
        # Hinglish / Romanized
        "ped", "vruksh", "nadi", "paani", "vayu", "mitti",
        "urja", "bijli", "kheti", "fasal", "pashu", "vanyajeev",
        "sanrakshan", "bachana", "harit", "swachh", "shuddh",
        "bhavishya", "aane wali peedhi", "hamari dharti",
        "duniya bachao", "insaan", "manav", "samaj",
    ],
}

def climate_relevance(text: str) -> int:
    t_lower  = text.lower()      # for case-insensitive Latin matching
    t_orig   = text              # for Devanagari (already case-irrelevant)
    score    = 0
    for kw in CLIMATE_KEYWORDS["high"]:
        if kw in t_lower or kw in t_orig:
            score += 3
    for kw in CLIMATE_KEYWORDS["medium"]:
        if kw in t_lower or kw in t_orig:
            score += 2
    for kw in CLIMATE_KEYWORDS["low"]:
        if kw in t_lower or kw in t_orig:
            score += 1
    return score

df["climate_score"] = df["text_clean"].apply(climate_relevance)

print(f"  Climate score distribution:")
print(f"    score = 0  : {(df['climate_score']==0).sum():,}  ← off-topic (fan comments, unrelated posts)")
print(f"    score >= 1 : {(df['climate_score']>=1).sum():,}")
print(f"    score >= 3 : {(df['climate_score']>=3).sum():,}")
print(f"    score >= 6 : {(df['climate_score']>=6).sum():,}")

before_rel  = len(df)
df          = df[df["climate_score"] >= MIN_CLIMATE_SCORE].reset_index(drop=True)
rel_removed = before_rel - len(df)

print(f"\n  Off-topic rows removed  : {rel_removed:,}")
print(f"  Climate corpus retained : {len(df):,}")
print(f"\n  Per-folder retention:")
for folder in df["folder"].unique():
    sub_final = (df["folder"] == folder).sum()
    sub_orig  = before["folders"].get(folder, 0)
    print(f"    {folder}: {sub_final:,} / {sub_orig:,} ({sub_final/sub_orig*100:.1f}%)")
```

```python
# ============================================================
# STAGE 11 — TRAIN / VAL / TEST SPLIT & EXPORT
# ============================================================
print("\n" + "="*60)
print(" STAGE 11 — Splitting & Exporting")
print("="*60)

df = df.sample(frac=1, random_state=42).reset_index(drop=True)

n        = len(df)
n_train  = int(n * TRAIN_RATIO)
n_val    = int(n * VAL_RATIO)

df_train = df.iloc[:n_train]
df_val   = df.iloc[n_train : n_train + n_val]
df_test  = df.iloc[n_train + n_val:]

EXPORT_COLS = ["text_clean", "source_file", "folder", "lang",
               "word_count", "climate_score"]

df_train[EXPORT_COLS].to_csv(f"{OUTPUT_DIR}/climate_train.csv",  index=False)
df_val[EXPORT_COLS].to_csv(  f"{OUTPUT_DIR}/climate_val.csv",    index=False)
df_test[EXPORT_COLS].to_csv( f"{OUTPUT_DIR}/climate_test.csv",   index=False)

with open(f"{OUTPUT_DIR}/climate_pretrain_corpus.jsonl", "w", encoding="utf-8") as f:
    for _, row in df.iterrows():
        record = {
            "text": row["text_clean"],
            "meta": {
                "source"        : row["source_file"],
                "region"        : row["folder"],
                "lang"          : row["lang"],
                "words"         : int(row["word_count"]),
                "climate_score" : int(row["climate_score"])
            }
        }
        f.write(json.dumps(record, ensure_ascii=False) + "\n")

print(f"  Train  : {len(df_train):,} rows  →  climate_train.csv")
print(f"  Val    : {len(df_val):,} rows   →  climate_val.csv")
print(f"  Test   : {len(df_test):,} rows   →  climate_test.csv")
print(f"  JSONL  : {n:,} rows   →  climate_pretrain_corpus.jsonl")
```

```python
# ============================================================
# STAGE 12 — BEFORE vs AFTER REPORT
# ============================================================
rows_dropped = before["total_rows"] - len(df)
pct_retained = 100 * len(df) / before["total_rows"]

print("\n")
print("┌──────────────────────────────────────────────────────────────┐")
print("│          CLIMATE DATASET — BEFORE vs AFTER                  │")
print("├───────────────────────────────────┬────────────┬────────────┤")
print("│ Metric                            │   BEFORE   │   AFTER    │")
print("├───────────────────────────────────┼────────────┼────────────┤")
print(f"│ Total rows                        │ {before['total_rows']:>8,}   │ {len(df):>8,}   │")
print(f"│ Duplicate texts                   │ {before['duplicate_texts']:>8,}   │ {0:>8,}   │")
print(f"│ Avg text length (chars)           │ {before['avg_text_len_chars']:>8}   │          —   │")
print(f"│ Avg word count                    │          —   │ {df['word_count'].mean():>8.1f}   │")
print(f"│ Avg climate score                 │          —   │ {df['climate_score'].mean():>8.2f}   │")
print(f"│ Unique languages                  │          —   │ {df['lang'].nunique():>8}   │")
print("├───────────────────────────────────┼────────────┼────────────┤")
print(f"│ TOTAL ROWS REMOVED                │            │ {rows_dropped:>8,}   │")
print(f"│ % CORPUS RETAINED                 │            │ {pct_retained:>7.1f}%   │")
print("├───────────────────────────────────┼────────────┼────────────┤")
print(f"│ Train split (90%)                 │            │ {len(df_train):>8,}   │")
print(f"│ Val split   ( 5%)                 │            │ {len(df_val):>8,}   │")
print(f"│ Test split  ( 5%)                 │            │ {len(df_test):>8,}   │")
print("└───────────────────────────────────┴────────────┴────────────┘")
print()
print(" REMOVAL BREAKDOWN:")
print(f"  ├─ Exact duplicates               : {exact_removed:,}")
print(f"  ├─ Near-duplicates                : {near_removed:,}")
print(f"  ├─ Fan/spam comments (NEW)        : {fan_removed:,}")
print(f"  ├─ Noisy (emoji/URL/char-spam)    : {noise_removed:,}")
print(f"  ├─ Too-short (< {MIN_WORDS} words)          : {length_removed:,}")
print(f"  ├─ Low quality (< 30% word-ratio) : {qual_removed:,}")
print(f"  └─ Off-topic (climate score = 0)  : {rel_removed:,}")
print()
print(" LANGUAGE DISTRIBUTION (final):")
for lang, cnt in df["lang"].value_counts().head(12).items():
    print(f"    {lang:>10} : {cnt:>6,}  ({cnt/len(df)*100:.1f}%)")
```
